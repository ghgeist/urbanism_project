```python
import os
from dotenv import load_dotenv

load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
# This works. I am able to connect to the neon db.
from langchain_community.utilities import SQLDatabase

db_uri = os.getenv("NEONDB_URI")
if db_uri is None:
    raise ValueError("NEONDB_URI environment variable is not set")

db = SQLDatabase.from_uri(db_uri)
print(db.dialect)
print(db.get_usable_table_names())
db.run("SELECT geoid20, natwalkind FROM national_walkability_index limit 1;")
```
